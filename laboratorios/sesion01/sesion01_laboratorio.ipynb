## SECCIÓN A: TRACK C
### Fase 1: El Artesano (15 min - Sin IA)
**Reto:** Inversión de buffer in-place (char*)

```python
#include <stdio.h>
#include <string.h>

void invertir(char *buffer) {
    if (buffer == NULL) {
        return;
    }

    char *inicio = buffer;
    char *fin = buffer + strlen(buffer) - 1;

    while (inicio < fin) {
        char temp = *inicio;
        *inicio = *fin;
        *fin = temp;

        inicio++;
        fin--;
    }
}

int main() {
    char texto[] = "hola";

    invertir(texto);

    printf("Resultado: %s\n", texto);

    return 0;
}
```

### Fase 2: El Científico (Prompting C-R-E-O)
Pega aquí la implementación optimizada propuesta por la IA:

```python
%%writefile ia_sesion01_c.c
#include <stdio.h>
#include <string.h>

// Implementación optimizada aplicando el contrato C-R-E-O
// Restricciones: O(n) tiempo, O(1) memoria, manejo seguro de punteros
void invertir_optimizada(char *buffer) {
    if (buffer == NULL || *buffer == '\0') {
        return;
    }

    char *inicio = buffer;
    char *fin = buffer + strlen(buffer) - 1;

    while (inicio < fin) {
        // Intercambio sin variable temporal adicional (XOR) o con registro temporal
        char temp = *inicio;
        *inicio = *fin;
        *fin = temp;

        inicio++;
        fin--;
    }
}

int main() {
    char texto[] = "estructura";
    invertir_optimizada(texto);
    printf("Optimizado: %s\n", texto);
    return 0;
}
```

### Fase 3: El Auditor (Test Harness en C)
Pruebas de estrés: Valores NULL, buffers vacios y longitudes impares

```python
%%writefile harness_sesion01_c.c
#include <stdio.h>
#include <string.h>
#include <assert.h>

// Importamos la lógica a auditar de manera directa
void invertir(char *buffer) {
    if (buffer == NULL || *buffer == '\0') {
        return;
    }
    char *inicio = buffer;
    char *fin = buffer + strlen(buffer) - 1;
    while (inicio < fin) {
        char temp = *inicio;
        *inicio = *fin;
        *fin = temp;
        inicio++;
        fin--;
    }
}

int main() {
    printf("=== INICIANDO AUDITORIA CRITICA: SESION 01 ===\n");

    // Caso 1: Puntero NULL (No debe tronar)
    printf("Probando caso: NULL... ");
    invertir(NULL);
    printf("PASO ✅\n");

    // Caso 2: Buffer vacío
    printf("Probando caso: Cadena vacía... ");
    char vacio[] = "";
    invertir(vacio);
    assert(strcmp(vacio, "") == 0);
    printf("PASO ✅\n");

    // Caso 3: Un solo carácter
    printf("Probando caso: Un carácter... ");
    char un_char[] = "A";
    invertir(un_char);
    assert(strcmp(un_char, "A") == 0);
    printf("PASO ✅\n");

    // Caso 4: Longitud impar
    printf("Probando caso: Longitud impar... ");
    char impar[] = "casas";
    invertir(impar);
    assert(strcmp(impar, "sasac") == 0);
    printf("PASO ✅\n");

    // Caso 5: Longitud par
    printf("Probando caso: Longitud par... ");
    char par[] = "hola";
    invertir(par);
    assert(strcmp(par, "aloh") == 0);
    printf("PASO ✅\n");

    printf("\n¡TODAS LAS PRUEBAS PASARON EXITOSAMENTE! 🎉✅\n");
    return 0;
}
```
